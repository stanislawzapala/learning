```python
import random
import json
import os

PLIK_STATYSTYK = "statystyki_losowan_stasimon.json"

def wczytaj_statystyki(sciezka):
    """Wczytuje historię losowań z pliku JSON."""
    if os.path.exists(sciezka):
        with open(sciezka, 'r', encoding='utf-8') as plik:
            return json.load(plik)
    return {}

def zapisz_statystyki(statystyki, sciezka):
    """Zapisuje zaktualizowaną historię losowań do pliku JSON."""
    with open(sciezka, 'w', encoding='utf-8') as plik:
        json.dump(statystyki, plik, ensure_ascii=False, indent=4)

def losuj_pytanie(czesc, umiane_klucze=None, plik_statystyk=PLIK_STATYSTYK, trudne=None):
    if trudne is None:
        trudne = []
    if umiane_klucze is None:
        umiane_klucze = set()
    else:
        umiane_klucze = set(umiane_klucze)
        
    dostepne_klucze = list(czesc.keys() - umiane_klucze)
    
    if not dostepne_klucze:
        return "Umiesz już wszystko z tej części!" 
        
    # 1. Wczytanie historii losowań (ile razy padło dane pytanie)
    statystyki = wczytaj_statystyki(plik_statystyk)
    
    # 2. Obliczenie wag dla dostępnych kluczy
    wagi = []
    for klucz in dostepne_klucze:
        # JSON zawsze zapisuje klucze słownika jako stringi, więc rzutujemy
        klucz_str = str(klucz)
        ile_razy_wylosowano = statystyki.get(klucz_str, 0)
        
        # Waga: 1 / (10 ^ liczba wylosowań). 
        # Pytanie wylosowane 0 razy ma wagę 1.0
        # Pytanie wylosowane 1 raz ma wagę 0.1
        # Pytanie wylosowane 3 razy ma wagę 0.001 itd.
        waga = 1.0 / (10 ** ile_razy_wylosowano)
        wagi.append(waga)
        
    # 3. Losowanie z uwzględnieniem wag
    # random.choices zwraca listę, więc bierzemy pierwszy (i jedyny) element [0]
    wylosowany_klucz = random.choices(dostepne_klucze, weights=wagi, k=1)[0]
    
    # 4. Aktualizacja i zapis statystyk
    klucz_str = str(wylosowany_klucz)
    statystyki[klucz_str] = statystyki.get(klucz_str, 0) + 1
    zapisz_statystyki(statystyki, plik_statystyk)
    
    if int(wylosowany_klucz) in trudne:
        print(f'Wylosowane pytanie jest w pytaniach trudnych! Przyłóż się!')
    return wylosowany_klucz, czesc[wylosowany_klucz]
```
