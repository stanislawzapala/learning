```python
def f2(names_raw, grades_raw):
    mk = {}
    for idx, name in enumerate(names_raw):
        klucze = mk.keys()
        if name not in klucze:
            mk[name] = [grades_raw[idx]]
        elif name in klucze:
            mk[name].append(grades_raw[idx])
    
    return mk
```

```python
f2(["Anna", "Jan", "Anna"], [3, 4, 5])
```

Projekt - pokazać (zakodować w pythonie) rozwiazanie jakiegoś problemu, który jest w artykule od Prowadzącego

```python
import matplotlib.pyplot as plt
import numpy as np

r = 3.8
def nasza_funkcja(x,r):
    return r*x*(1-x)

x_vals = np.linspace(0,1,1000)
y_vals = nasza_funkcja(x_vals, r)

fig, ax = plt.subplots(figsize=(8, 8))
    
ax.plot(x_vals, y_vals, 'k', lw=2, label=f'$f_r(x)$ dla r={r}')

ax.plot(x_vals, x_vals, 'k', lw=1.5, label='$y = x$')

x0 = 0.5

x = x0

for i in range(20):
    start_y = 0 if i == 0 else x    
    y = nasza_funkcja(x,r)
    ax.plot([x, x], [start_y, y], color='gray', lw=1.2)
    ax.plot(x, y, color='black', marker='o', markersize=6)
    
    ax.plot([x, y], [y, y], color='gray', lw=1.2)
    ax.plot(y, y, color='gray', marker='o', markersize=6, alpha=0.7)

    x = y

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_aspect('equal') 


plt.show()
```

```python
def nasza_funkcja_do_ntej_ostatnie10dodrugiejsposrod104(x,r,n):
    for i in range(n - 10**2):
        x = r*x*(1-x)
    
    last_x_vals = []
    for i in range(10**2):
        x = r*x*(1-x)
        last_x_vals.append(x)
    return last_x_vals
```

```python
x_vals = np.linspace(0,1,1000)
```

```python
r_values = np.linspace(2.5, 4.0, 1000)
for r in r_values:
    last_x_vals = nasza_funkcja_do_ntej_ostatnie10dodrugiejsposrod104(0.5, r, 1000)
    plt.plot([r]*len(last_x_vals), last_x_vals, 'k.', markersize=2)
```
