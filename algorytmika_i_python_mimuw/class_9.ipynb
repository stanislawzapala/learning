### Lista w zbiór

```python
def lista_w_zbior(x):
    zbior = []
    for i in x:
        if i not in zbior:
            zbior += [i]
    return zbior

# Złożoność pamięciowa: pesymistyczna O(n), optymistyczna O(1)
# Złożoność czasowa: pesymistyczna O(n^2), optymistyczna O(n)
```

```python
x = [1,2,3,4,4,4,4,4,5]
lista_w_zbior(x)
```

### Zadanie z prekolokwium

```python
def f2(l,t):
    zbior_do_sprawdzania_dopelnien = set(l)
    for idx, val in enumerate(l):
        dopelnienie_do_t = t-val
        if dopelnienie_do_t in zbior_do_sprawdzania_dopelnien:
            for i in range(idx+1, len(l)):
                if l[i] == dopelnienie_do_t:
                    return (idx, i)
    return None

# Tutaj źle, bo nie uwzględniamy sytuacji, że dopelnienie_do_t może być takie samo jak val, a wtedy musimy mieć co najmniej 2 takie same elementy w liscie, żeby zwrócić poprawny wynik.
```

```python
def f2_optymalne(l, t):
    widziane = {}  
    for idx, val in enumerate(l):
        dopelnienie = t - val
        if dopelnienie in widziane:
            return (widziane[dopelnienie], idx)
        widziane[val] = idx
    return None

# Złożoność czasowa: O(n) - przechodzimy listę tylko raz
# Złożoność pamięciowa: O(n) - w najgorszym razie zapiszemy całą listę w słowniku
```

```python
f2([1,2,3,4,4,4,5,6,6], 5)
```

### Inny sort

```python
def BC_sort(l):
    lista_b = []
    lista_c = []
    for i in l:
        if i == 'B':
            lista_b.append('B')
        elif i == 'C':
            lista_c.append('C')
    return lista_c + lista_b

# Czasowo: O(n)
# Pamięciowo: O(n)
```

### Mergesort

```python
def mergesort(l):
    tablica_pomocnicza = []
    if len(l) == 1:
        return l
    else:
        mid = len(l) // 2
        tab1 = mergesort(l[:mid])
        tab2 = mergesort(l[mid:])

        # Teraz scalamy
        i = 0
        j = 0
        while i < mid and j < len(l):
            if tab1[i] <= tab2[j]:
                tablica_pomocnicza.append(tab1[i])
                i += 1
            elif tab1[i] > tab2[j]:
                tablica_pomocnicza.append(tab1[j])
                j += 1
        while i < mid:
            tablica_pomocnicza.append(tab1[i])
            i += 1
        while j < len(l):
            tablica_pomocnicza.append(tab2[j])
            j += 1
        return tablica_pomocnicza
        
    
```

```python
tab = [1,3,2,5,4,1,2]
mergesort(tab)
```
