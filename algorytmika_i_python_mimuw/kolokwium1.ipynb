```python
def zad1(A,B):
    wynik = []
    setB = set(B)
    for a in A:
        if a in setB:
            wynik.append(a)
    return wynik

# Złożoność pamięciowa O(n+m), gdzie n i m to długości list A i B.
# Złożoność czasowa O(n+m), bo tworzymy zbiór z B w O(m) i potem sprawdzamy każdy element A w O(n) z O(1) czasem dostępu do zbioru.
```

```python
def zad2(l):
    average = sum(l) / len(l)
    wynik = []
    for i in l:
        if i > average:
            wynik.append(i)
    return wynik

# Złożoność pamięciowa O(n) w najgorszym przypadku, gdy wszystkie elementy poza jednym są większe od niego
# Złożoność czasowa O(n), ponieważ musimy przejść przez listę raz, raz do obliczenia średniej i raz do sprawdzenia, które elementy są większe od średniej.
```

```python
def zad3(t,x):
    i = len(t) // 2
    j = len(t) // 2 + 1

    while i >= 0 and j <= len(t):
        if t[i] + t[j] == x:
            return True
        elif t[i] + t[j] < x:
            j += 1
        else:
            i -= 1
    return False
```

```python
def zad4(s):
    suma = 0
    n = 0

    for val in s.values():
        n += 1
        suma += val
    
    srednia = suma / n
    przelewy = []

    for id1 in s.keys():
        if s[id1] > srednia:
            for id2 in s.keys():
                if s[id1] == srednia:
                    break
                elif s[id2] < srednia:
                    przelew = min(srednia-s[id2], s[id1]-srednia)
                    s[id1] -= przelew
                    s[id2] += przelew
                    przelewy.append((id1, id2, przelew))

    return przelewy
```

```python
slownik = {'1':150, '2':20, '3':40, '4':150}
zad4(slownik)
```
