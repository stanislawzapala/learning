# Iteracja i rekurencja

## Przykład: silnia

```python
def silnia_iter(n):
    silnia = 1
    for i in range(2, n+1):
        silnia *= i;
    return silnia;
```

```python
for i in range(7):
    print(f"{i}! = {silnia_iter(i)}");
```

```python
def silnia_rek(n):
    if(n==0):
        return 1;
        
    return n * silnia_rek(n-1);
```

```python
for i in range(7):
    print(f"{i}! = {silnia_rek(i)}");
```

```python
import math
```

```python
for i in range(7):
    print(f"{i}! = {math.factorial(i)}");
```

## Przykład: suma liczb z listy

```python
def suma(tab):
    suma = 0;
    for x in tab:
        suma += x;
    return suma;
```

```python
v = [9, 4, 2, 7, 11]
```

```python
suma(v)
```

## Przykład: znajdowanie minimum w tablicy (iteracyjnie)

```python
def min_iter(tab):
    if (len(tab)==0):
        return None;
    
    mini = tab[0];
    for x in tab[1:]:
        if(x < mini):
            mini = x;
    return mini;
```

```python
min_iter(v)
```

```python
min_iter([])
```

## Przykład: szukanie liczb pierwszych mniejszych niż podane n

```python
def czy_pierwsza(n):
    if(n<=1):
        return False;
    for i in range(2, n):
        if(n%i == 0):
            return False;
    return True;
```

```python
def czy_pierwsza(n):
    if(n<=1):
        return False;
    for i in range(2, int(n**0.5)+1):
        if(n%i == 0):
            return False;
    return True;
```

```python
for i in range(21):
    print(f"{i}: {czy_pierwsza(i)}");
```

```python
def pierwsze(n):
    if(n<=1):
        return [];

    if(czy_pierwsza(n)):
        return pierwsze(n-1) + [n];
    else:
        return pierwsze(n-1);
```

```python
pierwsze(50)
```

# Metoda dziel i rządź

```python
def min_dir(tab):
    n = len(tab);

    if(n==0):
        return None;
    elif(n==1):
        return tab[0];
    else:
        polowa = n//2;
        mini1 = min_dir(tab[:polowa]);
        mini2 = min_dir(tab[polowa:]);
        if(mini1<mini2):
            return mini1;
        else:
            return mini2;

# Złożoność czasowa: O(n)
# Złożoność pamięciowa: O(log n) - głębokość rekurencji ( choć tutaj inaczej, bo tworzymy nowe listy)
```

```python
v = np.random.normal(scale=20, size = 16).tolist()
print(v)
print(min_dir(v))
```

```python
min_dir([])
```

```python
min_dir([1])
```

## Przykład: sortowanie

```python
def scal_posortowane(tab1, tab2):
    polaczona = [];

    i1=0
    i2=0

    while True:
        if(i1 >= len(tab1)):
            polaczona += tab2[i2:];
            break;
        if(i2 >= len(tab2)):
            polaczona += tab1[i1:];
            break;
            
        if(tab1[i1] <= tab2[i2]):
            polaczona.append(tab1[i1]);
            i1 += 1
        else:
            polaczona.append(tab2[i2]);
            i2 += 1

    return polaczona;

# Złożoność czasowa: O(n+m), gdzie n,m to długości list
# Złożoność pamięciowa: O(n+m)
```

```python
scal_posortowane([1, 5, 7, 11, 77, 88], [2, 3, 8, 16, 32, 33])
```

```python
scal_posortowane([], [2, 3, 8, 16, 32, 33])
```

```python
scal_posortowane([2, 3, 8, 16, 32, 33], [])
```

```python
def sortuj_mergesort(tab):
    if(len(tab) < 2):
        return tab;

    polowa = len(tab)//2;
    tab1 = sortuj_mergesort(tab[:polowa]);
    tab2 = sortuj_mergesort(tab[polowa:]);

    return scal_posortowane(tab1, tab2);
    
```

```python
v = np.random.normal(scale=20, size = 17).tolist()
print(v)
print(sortuj_mergesort(v))
```
