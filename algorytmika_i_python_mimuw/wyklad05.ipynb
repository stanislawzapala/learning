# Algorytmy zachłanne

```python
def wydaj_reszte(ile_pln):
    co_wydac = [];
    zostalo = ile_pln;

    while(zostalo>0):
        for x in (500, 200, 100, 50, 20, 10, 5, 2, 1):
            if(zostalo - x >= 0 ):
                co_wydac.append(x);
                zostalo -= x;
                break; 

    return co_wydac;
```

# Szukanie przecięcia przedziałów

```python
def przeciecie(p, q):
    """
        Szukamy przecięcia otwartych przedziałów *p* i *q*. Zakładamy, że przedział to:

        * para liczb (a, b), odpowiadająca przedziałowi (a, b) dla a < b
        * None, oznaczający zbiór pusty

        Tak samo zwracamy wynik.
    """
    if (p is None) or (q is None):
        return None;

    assert len(p) == 2, "p nie jest parą 2 liczb"
    assert len(q) == 2, "q nie jest parą 2 liczb"
    assert p[0] < p[1]
    assert q[0] < q[1]

    r = ( max((p[0], q[0])), min((p[1], q[1])) )

    if (r[0] >= r[1]):
        return None;
    else:
        return r;

```

# NWD, algorytm Euklidesa

```python
def euklides(w, m):
    assert w>0;
    assert m>0;

    while (w != m):
        if(m > w):
            w, m = m, w;
        w -= m;
    return w;

# Złożoność pamięciowa: O(1)
# Złożoność czasowa: O(hmmmm nie wiem)
```

```python
euklides(-1, 5)
```

```python
for i, j in ( (1, 1), (7, 7), (1, 5), (5, 15), (25, 15), (27, 9), (13, 17), (27, 33) ):
    print(f"NWD({i}, {j}) = {euklides(i, j)}")
```

```python
def euklides2(w, m):
    assert w>0;
    assert m>0;
    
    if(m > w):
        w, m = m, w;
    
    while (m != 0):
        w, m = m, w%m;
        
    return w;
```

```python
for i, j in ( (1, 1), (7, 7), (1, 5), (5, 15), (25, 15), (27, 9), (13, 17), (27, 33) ):
    print(f"NWD({i}, {j}) = {euklides2(i, j)}")
```

# Obliczanie wartości wielomianu

```python
def wiel1(x, a):
    """
        Liczymy w(x) zgodnie z kolejnością działań.
    """
    assert len(a)>0, "a powinno być tablicą długości co najmniej 1"

    suma = 0;
    for i in range(len(a)):
        suma += a[i] * (x**i);
    return suma;
    
```

```python
wiel1(-1, [])
```

```python
a = [5]

for x in [0, 1, -1, 5.]:
    print(f"w({x}) = {wiel1(x, a)}")
```

```python
a = [-1, 1, 2]

for x in [0, 1, -1, 5.]:
    print(f"w({x}) = {wiel1(x, a)}")
```

```python
def wiel2(x, a):
    """
        Liczymy w(x) zgodnie z kolejnością działań.
    """
    assert len(a)>0, "a powinno być tablicą długości co najmniej 1"

    suma = a[0];
    for i in range(1, len(a)):
        suma += a[i] * (x**i);
    return suma;
    
```

```python
for a in [ [5], [-1, 1, 2]]:
    print(a)
    for x in [0, 1, -1, 5.]:
        print(f"w({x}) = {wiel2(x, a)}")
    print("-"*10)
```

```python
def wiel3(x, a):
    """
        Liczymy w(x) zgodnie z kolejnością działań.
    """
    assert len(a)>0, "a powinno być tablicą długości co najmniej 1"

    return sum([ a[i] * (x**i) for i in range(len(a))]);
```

```python
for a in [ [5], [-1, 1, 2]]:
    print(a)
    for x in [0, 1, -1, 5.]:
        print(f"w({x}) = {wiel3(x, a)}")
    print("-"*10)
```

## Metoda Hornera

```python
def horner(x, a):
    """
        Liczymy w(x) metodą Hornera.
    """
    assert len(a)>0, "a powinno być tablicą długości co najmniej 1"

    wynik = a[-1];

    for i in reversed(range(len(a)-1)):
        wynik = wynik*x + a[i];
    return wynik;
```

```python
for a in [ [5], [-1, 1, 2]]:
    print(a)
    for x in [0, 1, -1, 5.]:
        print(f"w({x}) = {horner(x, a)}")
    print("-"*10)
```

## Lista nieposortowana

```python
def szukaj1(x, lista):
    """
        Szukamy *x* na liście. Jeśli jest, zwracamy jego indeks, jeśli nie ma zwracamy *None*.
    """
    for i in range(len(lista)):
        if(lista[i] == x):
            return i;

    return None;
```

```python
l = np.random.randint(-30, 30, size=21).tolist()
l
```

```python
szukaj1(-13, l)
```

```python
def szukaj2(x, lista):
    """
        Szukamy *x* na liście. Jeśli jest, zwracamy jego indeks, jeśli nie ma zwracamy *None*.
    """
    for i, el in enumerate(lista):
        if(el == x):
            return i;

    return None;
```

```python
szukaj2(-13, l)
```

```python
szukaj2(8, l)
```
