```python
def func(x):
    count = 0
    for i in x:
        if i % 3 == 0:
            count += 1
    return count

# Złożoność czasowa: O(n), gdzie n to długość listy x
# Złożoność przestrzenna: O(1), ponieważ używamy stałej
```

```python
def f2(l):
    wynik = set()
    for t in l:
        for i in t:
            wynik.add(i)
    wynik_lista = list(wynik)
    return wynik_lista

# Złożoność czasowa: O(n*m), gdzie n to liczba elementów w liście l, a m to średnia liczba elementów w każdej podliście t
# Złożoność przestrzenna: O(k), gdzie k to liczba unikalnych elementów w podlistach t (w najgorszym przypadku k może być równa sumie wszystkich elementów w podlistach t)
# Optymistycznie pamięciowo jest O(1), jeśli wszystkie elementy w podlistach t są takie same, ale w ogólności jest O(k) ze względu na potrzebę przechowywania unikalnych elementów w zbiorze wynik.
```

```python
def f3(t, x):
    wynik = set()
    for i in range(len(t)):
        for j in range(i, len(t)):
            for k in range(j, len(t)):
                if t[i] + t[j] + t[k] == x:
                    wynik.add((i, j, k))
    return wynik
# Złożoność czasowa: O(n^3), gdzie n to długość listy t
# Złożoność przestrzenna: O(k), gdzie k to liczba unikalnych trójek (t[i], t[j], t[k]) takich, że t[i] + t[j] + t[k] == x (w najgorszym przypadku k może być równa liczbie wszystkich możliwych trójek, czyli 3 nad n), czyli O(n^3) w najgorszym przypadku, jeśli wszystkie trójki są unikalne.
```
