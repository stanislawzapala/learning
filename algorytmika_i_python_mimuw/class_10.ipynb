```python
def oblicz_an(n):
    if n == 0:
        return 1
    if n == 1:
        return 6
    if n == 2:
        return 4
        
    a0 = 1
    a1 = 6
    a2 = 4
    
    for i in range(3, n + 1):
        an = 2 * a2 + 5 * a1 - 6 * a0 + 12 * i - 38
        a0 = a1
        a1 = a2
        a2 = an
        
    return a2
```

```python
def moc_sumy_zbiorow(ciag1, ciag2):
    i = 0
    j = 0
    unikalne = 0
    ostatni = None
    
    while i < len(ciag1) and j < len(ciag2):
        if ciag1[i] < ciag2[j]:
            aktualny = ciag1[i]
            i += 1
        elif ciag1[i] > ciag2[j]:
            aktualny = ciag2[j]
            j += 1
        else:
            aktualny = ciag1[i]
            i += 1
            j += 1
            
        if aktualny != ostatni:
            unikalne += 1
            ostatni = aktualny
            
    while i < len(ciag1):
        if ciag1[i] != ostatni:
            unikalne += 1
            ostatni = ciag1[i]
        i += 1
        
    while j < len(ciag2):
        if ciag2[j] != ostatni:
            unikalne += 1
            ostatni = ciag2[j]
        j += 1
        
    return unikalne
```
