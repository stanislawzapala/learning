```python
def czy_wczesniej_lub_rowne(x, y):
    """
    Funkcja pomocnicza określająca, czy 'x' powinno stać przed 'y' (lub czy są równe)
    zgodnie z zasadami:
    1. Ujemne zawsze przed nieujemnymi.
    2. Jeśli oba są ujemne, sortujemy malejąco (np. -1 przed -5, czyli x >= y).
    3. Jeśli oba są nieujemne, sortujemy rosnąco (np. 1 przed 3, czyli x <= y).
    """
    if x < 0 and y >= 0:
        return True
    if x >= 0 and y < 0:
        return False
    if x < 0 and y < 0:
        return x >= y  # Ujemne malejąco (większa wartość ujemna jest wcześniej)
    return x <= y      # Nieujemne rosnąco (mniejsza wartość dodatnia jest wcześniej)

def podzial(l, i, j):
    p = i
    pivot = l[i] # Punkt podziału to początkowy element
    
    for q in range(i + 1, j + 1):
        # Zamiast l[q] <= l[i], korzystamy z naszej nowej logiki
        if czy_wczesniej_lub_rowne(l[q], pivot):
            p += 1
            l[p], l[q] = l[q], l[p]
            
    l[i], l[p] = l[p], l[i]
    return p

def quicksort(l, i=0, j=None):
    if j is None:
        j = len(l) - 1
        
    if i < j:
        p = podzial(l, i, j)
        quicksort(l, i, p - 1)
        quicksort(l, p + 1, j)
    
    return l # Opcjonalnie zwracamy listę dla wygody podglądu

# --- TESTOWANIE ---
przyklad = [-5, 3, 0, 1, -2, -1, 7, -7]
print("Przed:", przyklad)
quicksort(przyklad)
print("Po:   ", przyklad)
```

```python
def zad4(lista_raw):
    t = lista_raw[-1]
    l = lista_raw[0:len(lista_raw) - 1]

    print(t)
    print(l)
    
    dopelnienia = {}

    for idx1, i in enumerate(l):
        dop_i = t - i

        if dop_i in dopelnienia:
            return (dopelnienia[dop_i], idx1)
        else:
            dopelnienia[i] = idx1
    
    return None
```

```python
x = zad4([3, -1, 3, 6])
print(x)
```

```python
import numpy as np

def weirdsum():
    return np.round(np.sum([np.abs(np.sin(i + np.pi)) for i in range(1, 2026)]), 5)
```
