# H.I.85

### Złożoność czasowa
Wydaje się, że O(n^2). Pesymistycznie O(n^2), optymistycznie O(n).
### Złożoność pamięciowa
O(1)
### Co robi ten algorytm
Bubble sort imo


```python
def bubble_sort(a):
    n = len(a)
    for i in range(n-1):
        zm = False
        for j in range(n-1-i):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
                zm = True
        if not zm:
            return

# Czy to jest stabilny algorytm sortowania? Tak, ponieważ nie zmienia kolejności elementów o tej samej wartości.
# Czy to jest bubble sort? Tak, ponieważ jest to implementacja algorytmu bubble sort, który polega na porównywaniu sąsiednich elementów i zamienianiu ich miejscami, jeśli są w złej kolejności.
# Złożoność czasowa tego algorytmu to O(n^2) w najgorszym przypadku, O(n) w najlepszym przypadku (gdy lista jest już posortowana) i O(n^2) w średnim przypadku. Złożoność pamięciowa to O(1), ponieważ sortowanie odbywa się w miejscu (in-place).
```

```python
a = [5, 2, 9, 1, 5, 6]
bubble_sort(a)
print(a)
```

```python
import random

def generate_type(data_type, size = 30, data_range = (-10,20)):
    if data_type == 'list':
        data_type = [random.randint(a=data_range[0], b=data_range[1]) for _ in range(size)]
    elif data_type == 'tuple':
        data_type = tuple(random.randint(a=data_range[0], b=data_range[1]) for _ in range(size))
    elif data_type == 'set':
        data_type = set(random.randint(a=data_range[0], b=data_range[1]) for _ in range(size))
    elif data_type == 'dict':
        data_type = {i: random.randint(a=data_range[0], b=data_range[1]) for i in range(size)}
    elif data_type == 'frozenset':
        data_type = frozenset(random.randint(a=data_range[0], b=data_range[1]) for _ in range(size))
    
    return data_type
```

```python
print(generate_type('list'))
print(generate_type('tuple'))
print(generate_type('set'))
print(generate_type('dict'))
print(generate_type('frozenset'))
```
