```python
def zad4(a):
    i, j = 0, 1
    count = 0
    path = []
    next = None
    start = 0

    while i >= 0 and j <= len(a) - 1:
        if a[i] == a[j]:
            count += 2

            if i == start:
                path.append(i)
                i = j + 1
                j = i + 1
                start = i
            elif len(path) != 0:
                i = path.pop()
                j += 1
            else:
                i -= 1
                j += 1
        else:
            if next is not None:
                i = next
                j = i + 1
                next = None
            else:
                i+= 1
                j+=1
        if count == len(a):
            return True
    return False
```
