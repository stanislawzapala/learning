```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-4*np.pi, 4*np.pi, 5000)
plt.plot(x,np.sin(x), label='sin(x)')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Wykres funkcji sin(x)')
minima = np.arange(-3*np.pi, 4*np.pi, 2*np.pi)
maxima = np.arange(-2*np.pi, 4*np.pi, 2*np.pi)
plt.scatter(minima, np.sin(minima), color='red', label='Minima')
plt.scatter(maxima, np.sin(maxima), color='green', label='Maksima')
plt.legend()
plt.grid()
plt.show()
```
