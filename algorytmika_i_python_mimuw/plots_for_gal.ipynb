```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. WYKRESY W R^2 (2D) - Z HIPERBOLĄ
# ==========================================
def plot_r2():
    fig, ax = plt.subplots(figsize=(8, 8))
    
    # Zakres osi (zwiększamy trochę rozdzielczość siatki dla gładkości)
    x = np.linspace(-5, 5, 400)
    y = np.linspace(-5, 5, 400)
    X, Y = np.meshgrid(x, y)
    
    # A) Hiperpłaszczyzna w R^2 (Prosta: 2x - y - 1 = 0)
    y_line = 2 * x - 1
    ax.plot(x, y_line, color='blue', label='Prosta: 2x - y = 1')
    
    # B) Hiperpowierzchnia st. 2 w R^2 (Hiperbola: x^2 - y^2 = 1)
    # Przenosimy wszystko na jedną stronę: x^2 - y^2 - 1 = 0
    F_hyperbola = X**2 - Y**2 - 1
    
    # Rysujemy kontur dla wartości 0
    ax.contour(X, Y, F_hyperbola, levels=[0], colors='green')
    
    # Używamy pustego plotu (hack), aby dodać kontur do legendy
    ax.plot([], [], color='green', label='Hiperbola: x^2 - y^2 = 1')
    
    # Konfiguracja wyglądu osi
    ax.axhline(0, color='black', linewidth=1) # Oś X
    ax.axvline(0, color='black', linewidth=1) # Oś Y
    ax.set_xlim([-5, 5])
    ax.set_ylim([-5, 5])
    ax.grid(True, linestyle='--')
    ax.set_aspect('equal')
    ax.set_title('Przestrzeń R^2: Prosta i Hiperbola')
    ax.legend()
    
    plt.show()
# ==========================================
# 2. WYKRESY W R^3 (3D)
# ==========================================
def plot_r3():
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    
    # Zakres osi
    x = np.linspace(-5, 5, 50)
    y = np.linspace(-5, 5, 50)
    X, Y = np.meshgrid(x, y)
    
    # A) Hiperpłaszczyzna w R^3 (Płaszczyzna: x + y + z = 0 -> z = -x - y)
    Z_plane = -X - Y
    #ax.plot_surface(X, Y, Z_plane, alpha=0.5, color='cyan', label='Płaszczyzna')
    
    # B) Hiperpowierzchnia st. 2 w R^3 (Hiperboloida jednopowłokowa: x^2/4 + y^2/4 - z^2/4 = 1)
    # Przekształcamy wzór ze zdjęcia, by wyliczyć Z: z^2 = x^2 + y^2 - 4 -> z = +/- sqrt(x^2 + y^2 - 4)
    # Aby uniknąć błędów dziedziny (pierwiastek z liczby ujemnej), maskujemy wartości:
    Z_squared = X**2 - Y**2 - 1
    Z_squared[Z_squared < 0] = np.nan # Ukrywamy środek, gdzie hiperboloida "nie istnieje"
    
    Z_hyperboloid_pos = np.sqrt(Z_squared)
    Z_hyperboloid_neg = -np.sqrt(Z_squared)
    
    ax.plot_surface(X, Y, Z_hyperboloid_pos, alpha=0.7, color='orange')
    ax.plot_surface(X, Y, Z_hyperboloid_neg, alpha=0.7, color='orange')
    
    # Konfiguracja wyglądu
    ax.set_xlim([-5, 5])
    ax.set_ylim([-5, 5])
    ax.set_zlim([-5, 5])
    ax.set_xlabel('Oś X')
    ax.set_ylabel('Oś Y')
    ax.set_zlabel('Oś Z')
    ax.set_title('Przestrzeń R^3: Płaszczyzna i Hiperboloida jednopowłokowa')
    
    plt.show()

# ==========================================
# URUCHOMIENIE FUNKCJI
# ==========================================
if __name__ == '__main__':
    print("Generowanie wykresu dla R^2...")
    plot_r2()
    
    print("Generowanie wykresu dla R^3...")
    plot_r3()
```
