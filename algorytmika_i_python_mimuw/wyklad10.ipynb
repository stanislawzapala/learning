# Krótkie deklaracje list

```python
tab = []
for x in range(10):
    tab.append(x**3);
tab
```

```python
tab2 = [x**3 for x in range(10)]
tab2
```

```python
tab2 = [(-1)**x * x**3 for x in range(10)]
tab2
```

```python
tab2
```

```python
tab3 = [x for x in tab2 if x>0]
tab3
```

```python
# to nie krotka
kr = (x**3 for x in range(10))
kr
```

```python
kr = tuple(x**3 for x in range(10))
kr
```
