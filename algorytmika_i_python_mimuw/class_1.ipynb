# G.11

```python
def f(c0):
    i = 0
    while (c0 != 4) and (c0 != 2) and (c0 != 1):
        if (c0 % 2) == 0:
            c0 = c0 // 2
        else:
            c0 = 3 * c0 + 1
        i += 1
    return i

maks_kroki = 0
maks_start = 0

for n in range(1, 10000):
    kroki = f(n)
    if kroki > maks_kroki:
        maks_kroki = kroki
        maks_start = n

print(f"Najdłuższa trajektoria dla c0 = {maks_start}")
print(f"Wymaga kroków: {maks_kroki}")
```
