# H.I.14

```python
def f(t1, t2):
    result = tuple()
    for i in range(len(t1)):
        result += ((t1[i],t2[i]),)
    return result
```

```python
t1 = ('raz', 'dwa', 'trzy')
t2 = ('jeden', 'dwa', 'trzy')   

f(t1, t2)
```
